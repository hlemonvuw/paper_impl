# FMNIST 

```python
import torch
import torchvision
from torchvision.transforms import v2
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as nnFunctional

import matplotlib.pyplot as plt
import numpy as np
import random

SEED = 1234

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed(SEED)
torch.backends.cudnn.deterministic = True

device = torch.device(torch.accelerator.current_accelerator()
                      .type if torch.accelerator.is_available() else 'cpu')
print(device, 'device detected')


```

```python
transform = v2.Compose([
    v2.ToImage(),
    v2.ToDtype(torch.float32, scale=True)])
```

```python
batch_size = 600
train_fmnist = torchvision.datasets.FashionMNIST(root="./data", train=True, download=True, transform=transform)
train_fmnist_loader = torch.utils.data.DataLoader(train_fmnist, batch_size=batch_size,
                                          shuffle=True, num_workers=2)

test_fmnist = torchvision.datasets.FashionMNIST(root='./data', train=False,
                                       download=True, transform=transform)
test_fmnist_loader = torch.utils.data.DataLoader(test_fmnist, batch_size=batch_size,
                                         shuffle=False, num_workers=2)

print(f'Number of training examples: {len(train_fmnist)}')
print(f'Number of testing examples: {len(test_fmnist)}')
print(f'Number of train dataloader examples: {len(train_fmnist_loader)}')
print(f'Number of test dataloader examples: {len(test_fmnist_loader)}')
```
